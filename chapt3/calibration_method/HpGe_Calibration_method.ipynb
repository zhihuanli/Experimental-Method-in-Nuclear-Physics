# HPGe 探测器 $\gamma$ 能谱刻度方法

HPGe 能谱刻度包括能量刻度、峰宽（分辨率）刻度和 full-energy peak efficiency 刻度。三者都从峰拟合开始，但使用的拟合结果不同：峰中心用于能量刻度，$\sigma$ 用于峰宽刻度，信号分量的面积用于效率刻度。

## 能量刻度

能量刻度应覆盖实际分析使用的能区。常用标准源包括 $^{133}$Ba、$^{137}$Cs、$^{60}$Co、$^{152}$Eu 和 $^{22}$Na；也可使用能量经过可靠确认的天然本底线。峰能量和 gamma emission probability 应来自源证书或经过评价的核数据表。

对每条刻度线，在未刻度谱中拟合峰中心 $ch_i$，再将 $(ch_i,E_i)$ 写入刻度关系。先从一次函数开始，

$$E(ch)=a_0+a_1ch,$$

只有残差随能量出现有规律的弯曲时才加入二次项，

$$E(ch)=a_0+a_1ch+a_2ch^2.$$

阶数不是由“二次函数更完整”决定，而是由残差和独立检验峰决定。

### 峰形与局部本底

局部峰拟合统一写成

$$f(x)=s(x)+b(x),$$

其中 $s(x)$ 是 photopeak signal，$b(x)$ 是峰下的局部本底。孤立且近似对称的 photopeak 可从 Gaussian signal 和线性本底开始：

$$
s(x)=A\exp\!\left[-\frac{(x-\mu)^2}{2\sigma^2}\right],
\qquad b(x)=b_0+b_1x.
$$

若低能侧本底在峰附近形成平滑台阶，可把同一个本底分量写成

$$
b(x)=b_0+b_1x+
\frac{S}{2}\operatorname{erfc}\!\left(\frac{x-\mu}{\sqrt{2}\sigma}\right).
$$

`erfc` 项描述与 photopeak 相连的低能侧 step，不是另一个峰。若残差仍显示低能 tail，应改用相应的 tail model，而不是继续提高背景多项式阶数。峰中心、$\sigma$ 和峰面积都应从经过残差检验的同一个模型中取得。

能量刻度残差定义为

$$\Delta E_i=E_{i,\mathrm{ref}}-E_{\mathrm{cal}}(ch_i).$$

## 峰宽与能量分辨率

峰拟合得到的是未刻度坐标上的 $\sigma_{ch}$。应先用刻度曲线的局部斜率换算：

$$
\sigma_E=\left|\frac{dE}{dch}\right|\sigma_{ch},
\qquad
FWHM=2\sqrt{2\ln2}\,\sigma_E\approx2.355\sigma_E.
$$

HPGe 探测器常用下面的经验分辨率模型：

$$FWHM(E)=\sqrt{A+BE+CE^2}.$$

常数项、随能量线性增长的方差项和二次项可分别概括电子学噪声、载流子统计和电荷收集等贡献。它们在本作业中是经验参数；是否需要全部三项仍应通过残差判断。直接用普通二次多项式拟合 FWHM 可能在外推时给出负峰宽，不是合适的默认模型。

## Full-energy peak efficiency

在能量 $E_\gamma$ 处，full-energy peak efficiency 为

$$
\varepsilon(E_\gamma)=
\frac{N_{\mathrm{peak}}}
{A(t)\,P_\gamma\,t_{\mathrm{live}}},
\qquad
A(t)=A_0\,2^{-t/T_{1/2}}.
$$

$N_{\mathrm{peak}}$ 取自前面同一次局部拟合的信号分量 $s(x)$，本底分量 $b(x)$ 不计入峰面积。对等宽 histogram，若 `gaus` 的参数 $A$ 表示峰高、bin 宽为 $w$，则

$$N_{\mathrm{peak}}=\frac{A\sigma\sqrt{2\pi}}{w}.$$

峰拟合完成后，ROOT 返回参数 covariance matrix；面积误差传播使用 $A$、$\sigma$ 的误差及 $\operatorname{Cov}(A,\sigma)$。Covariance 是拟合结果，不是这里的拟合输入。若使用非 Gaussian signal，应直接积分所采用的 $s(x)$。Sideband subtraction 可以作为独立交叉检查，但不能在峰拟合中采用一种本底定义、计算面积时又换成另一种而不比较结果。

$A(t)$ 是测量时的源活度，$P_\gamma$ 是每次衰变发射该 gamma ray 的概率，$t_{\mathrm{live}}$ 是 live time。用 wall-clock time 代替 live time 只有在 dead time 可忽略或已修正时才成立。

<img src="full_energy_peak_area.png" alt="full-energy peak area above the fitted local background" style="max-width:78%;" />

橙色区域是总模型与同一个拟合本底之差，即用于效率计算的信号分量积分 $N_{\mathrm{peak}}$。

效率曲线通常在 log–log 坐标上拟合，例如在限定能区内使用低阶经验式

$$
\ln\varepsilon(E)=p_0+p_1\ln E+p_2(\ln E)^2.
$$

应检查相对残差，避免仅靠提高多项式阶数追随每一个点。源活度的不确定度会使同一核素的效率点相关；精密效率刻度还需考虑 dead time、true-coincidence summing、源的几何和自吸收等修正。缺少这些信息时，结果应称为未修正或 apparent efficiency，而不是最终绝对效率。
